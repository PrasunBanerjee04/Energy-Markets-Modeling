##### Shortcuts using the power modeling library PyPSA-USA 

```python
n = pypsa.Network()

for i in range(1, 5):
    n.add("Bus", f"Bus {i}")

branches = [("Bus 1", "Bus 2"), ("Bus 2", "Bus 3"), ("Bus 3", "Bus 4"), ("Bus 4", "Bus 1")]
for idx, (b0, b1) in enumerate(branches, start=1):
    n.add("Line", f"Line {idx}", bus0=b0, bus1=b1, x=0.1, r=0.0)

n.determine_network_topology()

for sub_network in n.sub_networks.obj:
    sub_network.calculate_PTDF() 
    
    print(f"PTDF Matrix for {sub_network.name}:\n", sub_network.PTDF)
```
